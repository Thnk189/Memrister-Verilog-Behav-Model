```python

p = 5  # The curve-fitting parameter

print("// Generated Joglekar Window Memory File")
print(f"// Parameters: p={p}, depth=256, width=8-bit")

# Formatted string AI helped out with this
# {:02X} = 2-digit uppercase Hex
# {:03} = 3-digit padded index
# {:.4f} = float rounded to 4 decimal places
fmt_string = "{:02X}  // Index {:03}: f(x)={:.4f}"

for i in range(256):
    # 1. Normalizer (0 to 1 range)
    x = i / 255.0

    # 2. Equation: f(x) = 1 - (2x - 1)^(2p)
    fx = 1.0 - (2.0 * x - 1.0)**(2 * p)

    # 3. Scale to 8-bit (0-255)
    scaled_val = int(round(fx * 255))

    # Ensure we don't exceed 8-bit limits (FF hex)
    if scaled_val > 255:
        scaled_val = 255

    # Create the formatted line
    output_line = fmt_string.format(scaled_val, i, fx)
    print(output_line)
```
